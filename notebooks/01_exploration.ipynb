```python
import pandas as pd
import os

RAW = "C:\\Users\\SUMANT\\Desktop\\ecommerce-sales-analysis\\data\\raw\\"

for fname in sorted(os.listdir(RAW)):
    if not fname.endswith(".csv"):
        continue
    df = pd.read_csv(RAW + fname)
    print(f"\n{'='*60}")
    print(f"FILE: {fname}")
    print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} cols")
    print(f"\nDtypes:\n{df.dtypes.to_string()}")
    print(f"\nNull counts:\n{df.isnull().sum().to_string()}")
    print(f"\nSample:\n{df.head(2).to_string()}")
```
